### Foundation Data Processing Functions

````python
# 🤖 AI Collaboration: Data Loading and Cleaning Function
# Ask AI to help you create a robust data loading function




import pandas as pd

def load_and_clean_transaction_data(file_path):

    try:
        # Load CSV file
        df = pd.read_csv(file_path)

        # Check required columns
        required_columns = ["Date", "Amount", "Category", "Description"]

        for column in required_columns:
            if column not in df.columns:
                raise ValueError(f"Missing required column: {column}")

        # Remove rows with missing values
        df = df.dropna()

        # Remove dollar signs
        df["Amount"] = (
            df["Amount"]
            .astype(str)
            .str.replace("$", "", regex=False)
            .str.replace(",", "", regex=False)
        )

        # Convert Amount column
        df["Amount"] = pd.to_numeric(df["Amount"])

        # Convert Date column
        df["Date"] = pd.to_datetime(df["Date"])

        # Remove duplicates
        df = df.drop_duplicates()

        # Reset index
        df = df.reset_index(drop=True)

        print("✅ Transaction data loaded and cleaned successfully")

        return df

    except FileNotFoundError:
        print("❌ File not found")

    except pd.errors.EmptyDataError:
        print("❌ CSV file is empty")

    except Exception as e:
        print(f"❌ Error loading data: {e}")


# Test your function
clean_data = load_and_clean_transaction_data(
    "sample_transactions.csv"
)

print(clean_data.head())
````

````python
# 🤖 AI Collaboration: Spending Analysis Function
# Ask AI to help you create comprehensive spending analysis

def analyze_spending_patterns(df):

    """
    Analyze spending patterns and generate business insights
    """

    # Calculate total spending
    total_spending = df["Amount"].sum()

    # Calculate totals by category
    category_totals = df.groupby("Category")["Amount"].sum()

    # Calculate category percentages
    category_percentages = (
        category_totals / total_spending
    ) * 100

    # Find top spending category
    top_category = category_totals.idxmax()

    # Store business insights
    insights = []

    # Generate insights
    if "Entertainment" in category_percentages:
        if category_percentages["Entertainment"] > 15:
            insights.append(
                f"Entertainment spending is high at {category_percentages['Entertainment']:.1f}% of total spending."
            )

    if "Groceries" in category_percentages:
        if category_percentages["Groceries"] > 40:
            insights.append(
                f"Groceries account for {category_percentages['Groceries']:.1f}% of spending. Meal planning may help reduce costs."
            )

    if "Coffee" in category_percentages:
        if category_percentages["Coffee"] > 10:
            insights.append(
                f"Coffee purchases represent {category_percentages['Coffee']:.1f}% of spending."
            )

    # Return analysis results
    return {
        "Total Spending": total_spending,
        "Category Totals": category_totals,
        "Category Percentages": category_percentages,
        "Top Category": top_category,
        "Insights": insights
    }


# Test your function
analysis = analyze_spending_patterns(clean_data)

print(analysis)
````

````python
# 🤖 AI Collaboration: Business Insights Generator
# Ask AI to help create professional financial recommendations


def generate_financial_recommendations(analysis_data):

    """
    Generate actionable financial recommendations
    based on spending analysis
    """

    # Get important analysis data
    top_category = analysis_data["Top Category"]
    total_spending = analysis_data["Total Spending"]
    insights = analysis_data["Insights"]

    # Create recommendation report
    report = f"""
===== Budget Buddy Financial Recommendations =====

Total Spending: ${total_spending:.2f}

Top Spending Category: {top_category}

Financial Insights:
"""

    # Add insights to report
    for insight in insights:
        report += f"- {insight}\n"

    # Add recommendations
    report += "\nRecommendations:\n"

    if top_category == "Groceries":
        report += "- Consider meal planning and shopping specials to reduce grocery costs.\n"
        report += "- Buy generic brands where possible.\n"

    elif top_category == "Entertainment":
        report += "- Reduce entertainment spending by setting a monthly budget.\n"
        report += "- Look for free or low-cost activities.\n"

    elif top_category == "Coffee":
        report += "- Making coffee at home may help reduce daily expenses.\n"
        report += "- Limit café visits during weekdays.\n"

    elif top_category == "Transport":
        report += "- Consider public transport or trip planning to reduce fuel costs.\n"

    else:
        report += "- Review your spending habits regularly.\n"

    # Add final advice
    report += "- Track weekly spending consistently.\n"
    report += "- Create a realistic monthly savings goal.\n"

    # Return formatted report
    return report


# Test your function
recommendations = generate_financial_recommendations(analysis)

print(recommendations)
````

## Custom Financial Tools

````python
# 🤖 AI Collaboration: Custom Tool Development
# Ask AI to help you create useful financial calculators



def create_savings_calculator_tool():
    """
    Create a custom savings goal calculator as an agent tool

    🤖 AI Collaboration Prompt:
    "Create a savings goal calculator function that takes current savings,
    monthly contribution, and target amount, then calculates time to reach
    goal. Format output for user-friendly display."
    """
    # Your AI-assisted tool implementation goes here
    # Include input validation, calculations, formatted output


def create_savings_calculator_tool():

    """
    Create a savings goal calculator tool
    """

    def savings_calculator(
        current_savings,
        monthly_contribution,
        savings_goal
    ):

        # Calculate remaining amount
        remaining_amount = savings_goal - current_savings

        # Prevent division by zero
        if monthly_contribution <= 0:
            return "Monthly contribution must be greater than 0."

        # Calculate months needed
        months_needed = remaining_amount / monthly_contribution

        # Generate formatted result
        result = f"""
======== Budget Buddy Savings Calculator ========

Current Savings: ${current_savings:.2f}

Monthly Contribution: ${monthly_contribution:.2f}

Savings Goal: ${savings_goal:.2f}

Remaining Amount: ${remaining_amount:.2f}

Estimated Time to Reach Goal:
{months_needed:.1f} months

=================================================
"""

        return result

    return savings_calculator


# Create tool
calculator = create_savings_calculator_tool()


# Test tool
result = calculator(
    1000,
    250,
    5000
)

print(result)
# Register your tool with the agent system
# agent.register_tool("savings_calculator", create_savings_calculator_tool)
````

## Gradio UI Integration

````python
# 🤖 AI Collaboration: Professional UI Design
# Ask AI to help you create a comprehensive Gradio interface

import gradio as gr

# Create chatbot response function
def finance_chat(user_question):

    question = user_question.lower()

    if "coffee" in question:
        return "Try making coffee at home more often to reduce daily spending."

    elif "grocery" in question:
        return "Creating a grocery budget and meal planning may help reduce costs."

    elif "save" in question:
        return "Setting automatic savings transfers can help build savings consistently."

    elif "entertainment" in question:
        return "Consider setting a monthly entertainment budget to control spending."

    elif "transport" in question:
        return "Using public transport more often may help reduce fuel expenses."

    else:
        return "Track your spending regularly and create a monthly budget."


# Create full Gradio UI
def create_finance_assistant_ui():

    # CSV analysis function
    def analyze_uploaded_csv(file):

        try:

            df = load_and_clean_transaction_data(
                file.name
            )

            analysis = analyze_spending_patterns(df)

            recommendations = generate_financial_recommendations(
                analysis
            )

            result = f"""
========= Uploaded CSV Preview =========

{df.head().to_string(index=False)}

========= Analysis =========

Total Spending: ${analysis['Total Spending']:.2f}

Top Category: {analysis['Top Category']}

========= Recommendations =========

{recommendations}
"""

            return result

        except Exception as e:

            return f"Error processing CSV file: {e}"

    # Create Gradio app
    with gr.Blocks() as demo:

        gr.Markdown(
            "# 💰 Budget Buddy Finance Assistant"
        )

        gr.Markdown(
            "Analyze spending, receive financial advice, and calculate savings goals."
        )

        # Chatbot tab
        with gr.Tab("Financial Chatbot"):

            chatbot_input = gr.Textbox(
                label="Ask a financial question"
            )

            chatbot_output = gr.Textbox(
                label="Budget Buddy Response"
            )

            chatbot_button = gr.Button(
                "Get Advice"
            )

            chatbot_button.click(
                fn=finance_chat,
                inputs=chatbot_input,
                outputs=chatbot_output
            )

        # Savings calculator tab
        with gr.Tab("Savings Calculator"):

            current_savings = gr.Number(
                label="Current Savings"
            )

            monthly_contribution = gr.Number(
                label="Monthly Contribution"
            )

            savings_goal = gr.Number(
                label="Savings Goal"
            )

            savings_output = gr.Textbox(
                label="Savings Result"
            )

            savings_button = gr.Button(
                "Calculate Savings Goal"
            )

            savings_button.click(
                fn=create_savings_calculator_tool(),
                inputs=[
                    current_savings,
                    monthly_contribution,
                    savings_goal
                ],
                outputs=savings_output
            )

        # CSV analysis tab
        with gr.Tab("CSV Spending Analysis"):

            csv_file = gr.File(
                label="Upload Transaction CSV"
            )

            csv_output = gr.Textbox(
                label="Financial Analysis",
                lines=20
            )

            csv_button = gr.Button(
                "Analyze Transactions"
            )

            csv_button.click(
                fn=analyze_uploaded_csv,
                inputs=csv_file,
                outputs=csv_output
            )

    return demo


# Launch application
demo = create_finance_assistant_ui()

demo.launch(share=True)
````
